# Task 1 :: Data Cleaning Challenge on Sales Data.

```python
import pandas as pd
```

```python
# Load Sales Data 
sales_2017 = pd.read_csv(r"C:\Users\gthangar\OneDrive - Nokia\1830_Automation\ML\Task\DataAnalysis\Sales Transactions-2017.csv")
sales_2018 = pd.read_csv(r"C:\Users\gthangar\OneDrive - Nokia\1830_Automation\ML\Task\DataAnalysis\Sales Transactions-2018.csv")
sales_2019 = pd.read_csv(r"C:\Users\gthangar\OneDrive - Nokia\1830_Automation\ML\Task\DataAnalysis\Sales Transactions-2019.csv")
```

```python
# Combine All Yearly Data
sales_data = pd.concat([sales_2017, sales_2018, sales_2019],ignore_index=True)
sales_data
```

```python
# Check number of rows and columns
sales_data.shape
```

```python
# Check column data types
sales_data.dtypes
```

```python
# Check missing values
sales_data.isna().sum()
```

```python
# Check duplicate rows
duplicate_count = sales_data.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove Duplicate Rows
sales_data.drop_duplicates(inplace=True)
sales_data
```

```python
# Remove Completely Empty Rows
sales_data = sales_data.dropna(how="all")
sales_data
```

```python
# Find rows where Date, Party and Product are all missing
missing_rows = sales_data[sales_data[["Date", "Party", "Product"]].isna().all(axis=1)]
print("Invalid transaction rows:", len(missing_rows))
```

```python
# Remove those rows
sales_data = sales_data.drop(index=missing_rows.index)
sales_data
```

```python
# Remove Rows with No Transaction Details
missing_rows = sales_data[sales_data[["Date", "Voucher", "Product", "Qty" ,"Rate","Gross","Disc","Voucher Amount"]].isna().all(axis=1)]
print("Rows with no transaction details:", len(missing_rows))
sales_data = sales_data.drop(index=missing_rows.index)
sales_data
```

```python
# Find rows where Date contains a blank space
empty_spaces = sales_data[sales_data["Date"].astype(str).str.strip() == ""]
print("Rows with blank Date:", len(empty_spaces))

# Remove those rows
sales_data = sales_data.drop(index=empty_spaces.index)
sales_data
```

```python
# Clean and Convert Date Column
sales_data["Date"] = pd.to_datetime(sales_data["Date"],format="%d/%m/%Y",errors="coerce")
sales_data
```

```python
# Convert Numeric Columns
numeric_columns = ["Qty","Rate", "Gross","Disc","Voucher Amount"]
for column in numeric_columns:
    sales_data[column] = pd.to_numeric(sales_data[column].astype(str).str.replace(",", "", regex=False).str.strip(),errors="coerce")
```

```python
sales_data.dtypes
```

```python
sales_data.isna().sum()
```

```python
# Handle Missing Discount Values
sales_data["Disc"] = sales_data["Disc"].fillna(0)
sales_data
```

```python
# Fill Voucher Amount Within Each Transaction
sales_data["Voucher Amount"] = (sales_data.groupby(["Date", "Voucher", "Party"])["Voucher Amount"].transform("first"))
sales_data
```

```python
# Remaining missing Voucher Amounts
sales_data["Voucher Amount"] = sales_data["Voucher Amount"].fillna(0)
sales_data
```

```python
# Check Missing Rate Values
missing_rate = sales_data[sales_data["Rate"].isna()]
print("Missing Rate rows:", len(missing_rate))
```

```python
# Calculate Missing Rate Values
sales_data.loc[sales_data["Rate"].isna(),"Rate"] = (sales_data["Voucher Amount"] + sales_data["Disc"]) / sales_data["Qty"]
# Calculate Missing Gross Values
sales_data.loc[sales_data["Gross"].isna(),"Gross"] = (sales_data["Qty"] * sales_data["Rate"])
```
